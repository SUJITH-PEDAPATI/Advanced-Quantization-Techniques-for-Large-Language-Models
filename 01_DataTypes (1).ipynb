# DataTypes
- Common DataTypes used to store the parameters of Machine Learning Models

```python
import torch
```

### Integer

```python
## Information of a 8-bit Unsigned Integer
torch.iinfo(torch.uint8)
```

```python
## Information of a 8-bit signed Integer
torch.iinfo(torch.int8)
```

### Floating Points

```python
## By default the Python stores the data in the fp64
value = 1/3
```

```python
format(value,'.60f')
```

```python
## 64 - bit floating points
torch_fp64 = torch.tensor(value,dtype = torch.float64)
```

```python
print(f"fp64 Tensor: {format(torch_fp64.item(),'.60f')}")
```

```python
tensor_fp32 = torch.tensor(value,dtype = torch.float32)
tensor_fp16 = torch.tensor(value,dtype = torch.float16)
tensor_bf16 = torch.tensor(value,dtype = torch.bfloat16)
```

```python
print(f"fp32 Tensor: {format(tensor_fp32.item(),'.60f')}")
print(f"fp16 Tensor: {format(tensor_fp16.item(),'.60f')}")
print(f"bf16 Tensor: {format(tensor_bf16.item(),'.60f')}")
```

```python
## Information of '16-bit brain floating point'
torch.finfo(torch.bfloat16)
```

```python
## Information of '32-bit floating point'
torch.finfo(torch.float32)
```
